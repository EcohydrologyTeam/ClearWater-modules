# History

[ClearWater-Modules](https://github.com/EcohydrologyTeam/ClearWater-modules) was envisioned as a collection of water quality process simulation modules that could easily be coupled to a variety of transport models.  These modules were developed by the [U.S. Army Engineer Research and Development Center (ERDC)](https://www.erdc.usace.army.mil/), [Environmental Laboratory (EL)](https://www.erdc.army.mil/EL/).

## V1

Version1 is a Python library that a python rewrite and modernization of the algorithms and structures originally written in Fortran 95, released as version 1.0 in 2021, and described in:

Zhang, Zhonglong and Billy E. Johnson. 2016. Aquatic nutrient simulation modules (NSMs) developed for hydrologic and hydraulic models. Vicksburg, MS: Environmental Laboratory, U. S. Army Engineer Research and Development Center (ERDC). Ecosystem Management and Restoration Research Program (EMRRP). ERDC/EL Technical Report 16-1. https://hdl.handle.net/11681/10112
Zhang, Zhonglong and Billy E. Johnson. 2016. Aquatic contaminant and mercury simulation modules developed for hydrologic and hydraulic models. Vicksburg, MS: Environmental Laboratory, U. S. Army Engineer Research and Development Center (ERDC). Environmental Quality Technology Research Program (EQTRP). ERDC/EL Technical Report 16-8. https://hdl.handle.net/11681/20249
Johnson, Billy E. and Zhonglong Zhang. 2016. Testing and Validation Studies of the NSMII-Benthic Sediment Diagenesis Module. Vicksburg, MS: Environmental Laboratory, U. S. Army Engineer Research and Development Center (ERDC). Ecosystem Management and Restoration Research Program (EMRRP). ERDC/EL Technical Report 16-11. https://hdl.handle.net/11681/20343

## V2 

Version2 is a signifcant refactor to address scalability and maintainability issues in the V1 version. The abstractions for the simulation framework have been simpilified and better aligned with the conceptional framework outlined in xarray-simlab. Model I/O has also been revisited to support chunked I/O to enable modeling with larger than memory datasets. Lastly variable management has been simplified with the addition of the VariableRegistry design pattern to enable modules to exchange data with zero copy operations. The V2 version is currently in active development and is not yet ready for release. 

# Installations

## Prerequisites

1. A python ecosystem. 
    The ClearWater modeling framework are a series of python packages that need to be installed our your system. Before any packages can be installed, you'll need a python environment on your system. If you are new to python we recommend that [Miniconda](https://www.anaconda.com/docs/getting-started/miniconda/main) environment management system. Follow their [install instructions](https://www.anaconda.com/docs/getting-started/miniconda/install) to get started.
1. The Git version control system.
    Git is a popular free and open source system that is used to managed different versions of files and is a foundational component of many software projects. To install the git system, we recommend you visit the [Git downloads page to get started](https://git-scm.com/downloads).
1. The clones of the core application repositories:
    In the Git system, files are collected and managed in a repository. The ClearWater modeling framework is currently distributed via a series of repositories available through [GitHub](https://github.com/EcohydrologyTeam/). You will need to clone these repositories locally so you have them available for use during the install steps.
    - [ClearWater-modules](https://github.com/EcohydrologyTeam/ClearWater-modules) contains the water quality process modules.
        - `gh repo clone EcohydrologyTeam/ClearWater-modules`
    - [ClearWater-riverine](https://github.com/EcohydrologyTeam/ClearWater-riverine) contains the 2D transport process modules. 
        - `gh repo clone EcohydrologyTeam/ClearWater-riverine`
    - [ClearWater-data](https://github.com/EcohydrologyTeam/ClearWater-data) is the common data engine for both models. 
        - `gh repo clone EcohydrologyTeam/ClearWater-data`

## Setup the Python Environment

1. Install the base python enviornment.
Navigate to the root directory of the ClearWater-modules repository and run the following command:
    ```bash
    conda env create --file environment.yml
    ```
2. Activate the environment.
    ```bash 
    conda activate clearwater
    ```
3. Install the three core ClearWater modules from there repositories using conda develop.
    ```bash
    conda develop path/to/ClearWater-modules/src/clearwater_modules_v2
    conda develop path/to/ClearWater-riverine/src
    conda develop path/to/ClearWater-data
    ```
