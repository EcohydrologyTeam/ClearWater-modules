# Background
Part of the refactor work has been to update the ClearWater modules code to utilize a [YAML](https://yaml.org) configuration file to define the model parameters and inputs. This better aligns with approaches familair to environmental models and the approach taken by [ClearWater-Riverine](https://github.com/EcohydrologyTeam/ClearWater-riverine).

# Example configuration
ClearWater Modules comes with an example configuration file built into package. 
This example file should demonstrate the basic structure of a configuration file 
and how to set up you own configuration file.
To output this configuration file, using either of the following:

In a python terminal session.
```python
    from clearwater_modules_v2.config import export_example
    export_example("optional_path/to/directory/or/file.yml")
```

In a commandline session with teh clearwater environment
```shell
    python -m clearwater_modules_v2.config.example "optional_path/to/directory/or/file.yml"
```
