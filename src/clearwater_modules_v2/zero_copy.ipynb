```python
#import dependencies
import numpy as np
import pandas as pd
import xarray as xr
from datetime import datetime
```

```python
#set up some data - in reality we'll load this from a file
minute_5 = xr.date_range("2025-01-01", "2025-01-02", freq="5min")
water_temperature = pd.DataFrame({'time': minute_5, 'water_temperature': np.ones(minute_5.size)})
```

```python
#put this in a Xarray Dataset
data = xr.Dataset(
    {
        "water_temperature": xr.DataArray(
            water_temperature["water_temperature"].values,
            coords={"time": water_temperature["time"]},
            dims=("time"),
        ),
    },
    coords={"time": minute_5}
)
```

```python
#We'll have a registry that handles this behavior, but we can mock it with a dictionary for now
registry = {}
registry['water_temperature'] = data["water_temperature"]
```

```python
#ok now have our registry let's verify behaviors
timestep = datetime(2025, 1, 1,0,0)
```

```python
#I should be able to access the timestep slice using the sel command
registry["water_temperature"].sel(time=timestep)
```

```python
# Now I should be able to modify that values
water_temperature = registry["water_temperature"].sel(time=timestep)
water_temperature *= 2
water_temperature
```

```python
#what I want is to have water_temperature in the original (registry dataarray) to have be modified
water_temperature_registry = registry["water_temperature"].sel(time=timestep)
water_temperature_registry
```
